```python
# ollama embeddings
# ollama supports embeddding models making it possible to build retrieval-augmented generation (RAG) applications locally.
# application that combine text promt with existing documents or other data
from langchain_community.embeddings.ollama import OllamaEmbeddings
embeddings = OllamaEmbeddings(model="gemma:2b")
```

```python
r1 = embeddings.embed_documents(
    [
        'alpha is the first letter of greek alphabet',
        'beta is the second letter of greek alphabet',
    ]
)
```

```python
# Huggingface embeddings
import os
from dotenv import load_dotenv
load_dotenv() # load all the variables from .env file
os.environ["HUGGINGFACE_API_KEY"] = os.getenv("HUGGINGFACE_API_KEY")
```

```python
# install sentence-transformers
from langchain_community.embeddings import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name = "all-MiniLM-L6-v2")# it will convert to 384 dimension vector
```

```python
text = "This is a test"
vector = embeddings.embed_query(text)
print(vector)
```
