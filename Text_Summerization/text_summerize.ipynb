```python
# to load the environment variables from a .env file
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
# to get the Groq API key from environment variables
groq_api_key = os.getenv("GROQ_API_KEY")
# to initialize llm with Groq
from langchain_groq import ChatGroq
llm = ChatGroq(api_key=groq_api_key, model="llama-3.3-70b-versatile")
llm
```

```python
# import necessary modules
from langchain.schema import (
    AIMessage,
    HumanMessage,
    SystemMessage   
)
```

```python
speech = """
My dear citizens of India,
Today, I stand before you with pride and hope as our nation steps into a new era of progress. India has always been a land of courage, culture, and intelligence. From the ancient days of Nalanda and Takshashila to our modern space missions, we have consistently shown the world the power of Indian determination.

Our government believes that development should reach every corner of India. Whether a child is born in a small village in Odisha or in a big city like Mumbai, the opportunity for growth must be the same. Our priority is to build strong foundations — better roads, clean drinking water, more hospitals, smart schools, and reliable electricity in every home.

In the last few years, technology has become India’s biggest partner. With digital services, citizens no longer need to stand in long lines for essential tasks. Aadhaar, UPI payments, online education, and e-governance systems are helping people live easier, faster, and more secure lives. This is not just modernization; it is a movement towards equality.

We are also supporting our youth — the strength of our nation. Programs focusing on skill development, startups, and innovation are helping young Indians convert their ideas into successful careers. Our vision is not just to create jobs, but to create employers who will take India to global leadership in industries like artificial intelligence, space technology, and renewable energy.

Agriculture remains the backbone of our nation. We are bringing more technology to farms, ensuring farmers get a fair price for their produce and better access to markets. Our goal is to transform farming into a profitable, respected profession, not a struggle for survival.

India’s role in global peace and climate protection is stronger than ever. We believe that sustainable development is the only path forward. New investments in green energy, electric mobility, and forest conservation will protect our environment for future generations. Our earth is a shared home, and India will always be a responsible guardian of nature.

Unity is our greatest strength. India is a country of many languages, religions, cultures, and traditions — yet we stand together as one. Discrimination has no place in the India we are building. Every citizen, regardless of background, deserves dignity and opportunity.

As we move toward the vision of a Developed India by 2047, every small step counts. The contribution of a teacher, the creativity of a student, the hard work of a farmer, the service of a soldier — all these form the soul of our progress. Let us promise to work together, think big, and keep India first in everything we do.

My dear friends, the future is full of opportunities. Let us embrace change, innovate with courage, and lead the world with confidence. The India of tomorrow is not just a dream — it is a destination we will reach together.
Jai Hind! Jai Bharat!
"""
```

```python
# define the messages for the chat model
chat_messages = [
    SystemMessage(content="You are a helpful assistant that summarizes speeches."),
    HumanMessage(content=f"Summarize the following speech in concise points:\n\n{speech}")
]
```

```python
# to get the how many tokens are used
llm.get_num_tokens(speech)  # integer value, to get we have to pass string input
```

```python
# one way of getting the response from the llm -- first way of text summerization
response = llm(chat_messages)# AIMessage object
response.content
```

## Prompt Template Text Summerization

```python
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain

# define the prompt template for text summerization
generic_template = """
Write the summary of the following speech in concise points:
{speech}
Translate the precise summary to {language}.
"""
prompt = PromptTemplate(
    input_variables=["speech", "language"],
    template=generic_template
)
prompt
```

```python
# to form the complete prompt
final_prompt = prompt.format(speech=speech, language="Kannada")
final_prompt
```

```python
# to get the tokens of the final prompt
llm.get_num_tokens(final_prompt)  # integer value, to get we have to pass string input
```

initial speech contain 598, but final prompt conain 621 because i added some more texts to speech which is " generic_template"

```python
# to create a llm chain for text summerization using prompt template
chain = LLMChain(llm=llm, prompt=prompt)
summary = chain.run(speech=speech, language="Kannada")
summary
```

## StuffDocumentChain Text Summerization

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("drapj.pdf")
docs = loader.load_and_split()
docs
```

```python
# Prompt template for text summerization using StuffDocumentChain
from langchain.prompts import PromptTemplate
prompt_template = """
Write short summary of the following speech in concise points,
speech: {text}
"""
prompt = PromptTemplate(
    input_variables=["text"],
    template=prompt_template
)
```

```python
from langchain.chains.summarize import load_summarize_chain
chain = load_summarize_chain(llm, chain_type="stuff", prompt=prompt,verbose=True)# first it will combine all the documents into one and then will pass prompt to llm
summary = chain.run(docs)
summary
```

## MapReduce to Summerize Large Documents

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
docs
```

```python
final_documents = RecursiveCharacterTextSplitter(chunk_size=2000, chunk_overlap=100).split_documents(docs)
final_documents
```

```python
# define the prompt template for text summerization using map_reduce chain
from langchain.prompts import PromptTemplate
prompt_template = """
Write short summary of the following speech in concise points,
speech: {text}
"""
map_prompt = PromptTemplate(
    input_variables=["text"],
    template=prompt_template
)
```

```python
# final prompt for text summerization using map_reduce chain
final_prompt = """
Provide the final summary based on the following important points:
Add a motivation title , start the precise summery with an introduction and provide the summery
in number of points for the speech.
speech: {text}
"""
final_prompt=PromptTemplate(
    input_variables=["text"],
    template=final_prompt
)
```

```python
# summery chain using map_reduce
from langchain.chains.summarize import load_summarize_chain
chain = load_summarize_chain(
    llm, 
    chain_type="map_reduce", 
    map_prompt=map_prompt, 
    combine_prompt=final_prompt, 
    verbose=True
)
summary = chain.run(final_documents)
summary
```

# RefineChain Text Summerization

```python
# RefineChain Text Summerization

chain = load_summarize_chain(
    llm,
    chain_type="refine",
    verbose=True
)
summary = chain.run(final_documents)
summary
```
