#### Simple LLM Using LCEL(LangChain Expression Language)
This application will translate text from English to other language. It is single llm call with prompting. 

```python
import os
from dotenv import load_dotenv
load_dotenv()

# to get openai api key
# import openai
# openai.api_key = os.getenv("OPENAI_API_KEY")

# to load groq api key
groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
# pip install langchain_groq --> interact with groq models and deployments

from langchain_groq import ChatGroq
model = ChatGroq(model="Gemma2-9b-It", groq_api_key=groq_api_key)
model
```

```python
# pip install langchain_core
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage(content="You are a helpful assistant that translates English to French."),# instruct llm model to how it needs to behave
    HumanMessage(content="How are you?"),
]
result = model.invoke(messages)
```

```python
# to display only output response from the model
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
output_parser.invoke(result)
```

```python
# Using LCEL we can chain the components together to build a simple app

chain = model | output_parser
chain.invoke(messages)
```

```python
# Prompt templates
from langchain_core.prompts import ChatPromptTemplate

generic_template = "Translate the following text into {language}:"

prompt = ChatPromptTemplate.from_messages(
    [
        ("system",generic_template),
        ("human","{text}")
    ]
)
result = prompt.invoke({"language":"French","text":"Hello"})
```

```python
result.to_messages()
```

```python
chain = prompt | model | output_parser
chain.invoke({"language":"French","text":"Hello"})
```
