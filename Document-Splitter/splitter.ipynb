```python
# Web Page Loader
from langchain_community.document_loaders import WebBaseLoader
loader = WebBaseLoader(web_paths=('https://en.wikipedia.org/wiki/Artificial_intelligence',))
web_documents = loader.load()# Load web page documents from the specified URL
web_documents # here we get all the text from the webpage
```

```python
# How to recursively split text by characters
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
)
final_docs = text_splitter.split_documents(web_documents)# to split documents into smaller chunks
final_docs
print(type(final_docs[0]))
```

```python
speech = ""
with open('speech.txt','r') as f:
    speech = f.read()

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
)
final_docs = text_splitter.create_documents([speech])# to create documents from raw text
final_docs
```

```python
type(final_docs[0])
```

```python
# Character Text Splitter
from langchain.text_splitter import CharacterTextSplitter
text_splitter = CharacterTextSplitter(
    separator="\n",
    chunk_size=500,
    chunk_overlap=50,
    length_function=len,
)
texts = text_splitter.split_documents(final_docs)# to split raw text into smaller chunks
texts
```

```python
# HTML Text-Splitter
from langchain_text_splitters import HTMLHeaderTextSplitter

html_string = """<html>
<head>
    <title>Sample HTML Document</title>
</head>
<body>
    <h1>Main Title</h1>
    <p>This is a sample paragraph under the main title.</p>
    <h2>Subheading 1</h2>
    <p>This is a sample paragraph under subheading 1.</p>
    <h3>Sub-subheading 1.1</h3>
    <p>This is a sample paragraph under sub-subheading 1.1.</p>
    <h2>Subheading 2</h2>
    <p>This is a sample paragraph under subheading 2.</p>
    <h3>Sub-subheading 2.1</h3>
    <p>This is a sample paragraph under sub-subheading 2.1.</p>
    <h4>Sub-sub-subheading 2.1.1</h4>
    <p>This is a sample paragraph under sub-sub-subheading 2.1.1.</p>
</body>
</html>
"""
header_to_split_on = [
    ("h1","Header 1"),
    ("h2","Header 2"),
    ("h3","Header 3"),
    ("h4","Header 4")
]
html_splitter = HTMLHeaderTextSplitter(header_to_split_on)
html_header_splits = html_splitter.split_text(html_string)# to split HTML content based on header tags
html_header_splits
```

```python
# using url
url = "https://arxiv.org/abs/1706.03762"
html_splitter = HTMLHeaderTextSplitter(header_to_split_on)
html_header_splits = html_splitter.split_text_from_url(url)# to split HTML content from a URL based on header tags
html_header_splits
```

```python
# JSON Text Splitter
import json
import requests

url = "https://jsonplaceholder.typicode.com/posts"
json_data = requests.get(url).json()# Fetch JSON data from the specified URL
json_data
```

```python
from langchain_text_splitters import RecursiveJsonSplitter

# Wrap the list in a dictionary
json_data_wrapped = {"posts": json_data}

json_splitter = RecursiveJsonSplitter()
json_splits = json_splitter.split_json(json_data_wrapped)# to split JSON data into smaller chunks
for split in json_splits[:2]:
    print(split)
    print("\n---\n")
```
