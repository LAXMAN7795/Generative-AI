### Streaming and Batch
#### Streaming

##### Most models can stream their output content while it is generated. By displaying output progressively, streamingimproves user experience , perticulerly for longer responses. Calling stream() returns an iterator that yeilds ouput chunks as they are produced. You can use a loop to process each chunk in real time.

```python
for chunk in model.stream("what is Generative AI"):# no need to wait for completion of generation of text, after generating instantly print the contents, but in .invoke() we have to wait for complete generatjion of the texts.
    print(chunk.text, end="|",flush=True)
```

### Batch: Collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel.

```python
response = model.batch([
    "How do parrots have colorful feathers?",
    "How do airoplanes fly?",
    "What is artificial intelligence"
],config={
    'max_concurrecy':5, # limit to 5 parellel call, if 10 questions are passed then 5 and 5 batches are passed to llm
})
for res in response:
    print(res.text)
```
